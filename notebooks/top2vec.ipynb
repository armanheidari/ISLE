```python
from top2vec import Top2Vec
from wordcloud import WordCloud
import matplotlib.pyplot as plt
```

```python
input_path = '../data/datasets/corpus.txt'
with open(input_path, 'r', encoding='utf-8') as f:
    loaded_sentences = [line.strip() for line in f if line.strip()]

print(f'Loaded {len(loaded_sentences)} sentences')
print(loaded_sentences[:10])
```

```python
model = Top2Vec(loaded_sentences, embedding_model='universal-sentence-encoder')
```

```python
# Get number of topics found
num_topics = model.get_num_topics()
print(f"Number of topics: {num_topics}")

# Get the topics
topic_words, word_scores, topic_nums = model.get_topics()

for topic, words, scores in zip(topic_nums, topic_words, word_scores):
    freq = dict(zip(words, scores))
    wc = WordCloud(width=800, height=400, background_color='white')
    wc.generate_from_frequencies(freq)
    plt.figure(figsize=(10, 5))
    plt.imshow(wc, interpolation='bilinear')
    plt.title(f"Topic {topic} Word Cloud")
    plt.axis('off')
    plt.show()
```
